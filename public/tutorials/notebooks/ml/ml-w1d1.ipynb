Today's code is short: load a real dataset, look at **features** and **labels**,
and see the same data the way a *supervised* and an *unsupervised* learner would.

Run each cell with **Shift + Enter**.

```python
from sklearn.datasets import load_iris

iris = load_iris()
X, y = iris.data, iris.target        # X: features, y: labels

print("X shape:", X.shape)            # 150 flowers, 4 measurements each
print("features:", iris.feature_names)
print("labels:", iris.target_names)
print("first flower:", X[0], "->", iris.target_names[y[0]])
```

Each **row** of `X` is one example (one flower). Each **column** is one feature.
`y` holds the answer for every row: 0 = setosa, 1 = versicolor, 2 = virginica.

Now let's draw two features. On the left we colour by label (what a **supervised** learner gets).
On the right the labels are hidden (what an **unsupervised** learner gets).

```python
import matplotlib.pyplot as plt

fig, (left, right) = plt.subplots(1, 2, figsize=(9, 3.8), sharex=True, sharey=True, layout="constrained")
for k, name in enumerate(iris.target_names):
    pick = y == k
    left.scatter(X[pick, 2], X[pick, 3], s=36, label=name)
left.set_title("Supervised view: labels given")
left.legend(loc="upper left")

right.scatter(X[:, 2], X[:, 3], s=36, color="#bcb7cb")
right.set_title("Unsupervised view: no labels")

for ax in (left, right):
    ax.set_xlabel("petal length (cm)")
left.set_ylabel("petal width (cm)")
plt.show()
```

Even without labels you can *see* groups on the right. Finding them automatically is
what clustering does (week 6).

## A model is a function with a knob
Here is a tiny hand-made classifier with one knob, `threshold`.
"Learning" means choosing the knob from data.

```python
import numpy as np

def tiny_model(petal_length, threshold):
    """Predict 1 (setosa) when the petal is short, else 0."""
    return (petal_length < threshold).astype(int)

is_setosa = (y == 0).astype(int)
for threshold in [1.5, 2.5, 4.0]:
    accuracy = np.mean(tiny_model(X[:, 2], threshold) == is_setosa)
    print(f"threshold = {threshold:.1f} cm  ->  accuracy = {accuracy:.0%}")
```

## Solution
Try the exercise yourself first: it's fine to get stuck for a minute. Then compare with this one (there are many good answers).

```python
perfect = []
for threshold in np.round(np.arange(1.0, 6.01, 0.1), 1):
    accuracy = np.mean(tiny_model(X[:, 2], threshold) == is_setosa)
    if accuracy == 1.0:
        perfect.append(threshold)

print(f"{len(perfect)} perfect thresholds, from {min(perfect):.1f} cm to {max(perfect):.1f} cm")
print(f"The longest setosa petal is {X[y == 0, 2].max()} cm and the shortest of the others is "
      f"{X[y != 0, 2].min()} cm: any threshold in that gap separates them.")
```
