Amplitudes are complex numbers: arrows with a length and a phase.
The lengths give the 0/1 odds; the phases show up when we compare states.

```python
import numpy as np

z = 0.6 * np.exp(1j * np.pi / 3)          # length 0.6, phase 60 degrees
print("z          =", np.round(z, 3))
print("length |z| =", round(abs(z), 3))
print("phase      =", round(np.degrees(np.angle(z)), 1), "degrees")
print("|z|^2      =", round(abs(z) ** 2, 3), " <- the only part the Born rule uses")
```

## Same odds, different states

```python
from qiskit.quantum_info import Statevector

s = 1 / np.sqrt(2)
states = {
    "|0>": Statevector([1, 0]),
    "|1>": Statevector([0, 1]),
    "|+>": Statevector([s, s]),
    "|->": Statevector([s, -s]),
    "|+i>": Statevector([s, 1j * s]),
}
for name in ["|+>", "|->", "|+i>"]:
    print(f"{name:>4}: amplitudes {np.round(states[name].data, 3)}  P(0), P(1) = {states[name].probabilities().round(2)}")
```

## Every overlap $|\langle\phi|\psi\rangle|^2$
1 means identical, 0 means perfectly distinguishable.

```python
names = list(states)
print("        " + "".join(f"{n:>7}" for n in names))
for a in names:
    row = [abs(np.vdot(states[a].data, states[b].data)) ** 2 for b in names]
    print(f"{a:>6}  " + "".join(f"{v:7.2f}" for v in row))
```

## A global phase changes nothing
`equiv` compares two states while ignoring a global phase.

```python
psi = Statevector([0.6, 0.8j])
turned = Statevector(np.exp(1j * 1.234) * psi.data)       # turn both arrows by 1.234 radians
print("same amplitudes?    ", np.allclose(psi.data, turned.data))
print("same probabilities? ", np.allclose(psi.probabilities(), turned.probabilities()))
print("same physical state?", psi.equiv(turned))
print("same overlap with |+>?", np.isclose(abs(np.vdot(states["|+>"].data, psi.data)) ** 2,
                                            abs(np.vdot(states["|+>"].data, turned.data)) ** 2))
```

## Solution

```python
overlap_plus = abs(np.vdot(plus, psi)) ** 2
overlap_plus_i = abs(np.vdot(plus_i, psi)) ** 2
print(f"|<+|psi>|^2  = {overlap_plus:.2f}")
print(f"|<+i|psi>|^2 = {overlap_plus_i:.2f}")
print("By hand: <+i|psi> = (0.6 + (-i)(0.8i)) / sqrt(2) = 1.4 / sqrt(2), and 1.4^2 / 2 = 0.98.")
print("So psi is almost |+i>, even though its 0/1 odds are 36/64.")
```
